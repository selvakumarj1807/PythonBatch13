# While Loop

```python
i = 0 # Value Initial

n = 10

while i <= n: # Value Condition #  11 <= 10
    print(i)

    i = i + 1 # Value Increment
```

```python
i = 0 

n = 10

while i <= n:
    print(i, end=', ')

    i = i + 1 
```

# Even Number

```python
i = 0 

n = 10

while i <= n: 
    print(i)

    i = i + 2 
```

# Odd Number

```python
i = 1 

n = 10

while i <= n: 
    print(i)

    i = i + 2 
```

# Jumping Statements

# Break

```python
i = 0 

n = 10

while i <= n:
    print(i)

    if i == 5:
        i = i + 1 
        print('Before Break')
        break
        print('After Break')

    i = i + 1 
```

# Continue

```python
i = 0 

n = 10

while i <= n:
    print(i)

    if i == 5:
        i = i + 3 
        print('Before Continue')
        continue
        print('After Continue')

    i = i + 1 
```

# Pass

```python
i = 0 

n = 10

while i <= n:
    print(i)

    if i == 5:
        i = i + 3 
        print('Before Pass')
        pass
        print('After Pass')

    i = i + 1 
```

# Without Condition 

```python
i = 0 

valueCondition = True

while valueCondition:
    print(i)

    if i == 5:
        i = i + 1 
        print('Before False')
        valueCondition = False
        print('After False')

    i = i + 1 
```
