# DataType & Variables

# DataTypes

int - 0 to 9

float - 0.5, 1.5,....

str - a to z, A to Z, Special Charecters(!@#$%^&*)

bool - True, False

```python
# Variables

userName = 'Sara'

age = 25

avg = 90.5

married = True

print(userName, type(userName))

print(age, type(age))

print(avg, type(avg))

print(married, type(married))
```

# Concatination

```python
userName = 'Sara'

age = 25

print('Hi', userName, '!...', 'your age is', age)
```
