# Date Time Module

```python
import datetime as dt
```

# Date

```python
currentDate = dt.date.today()

print(currentDate)
```

```python
userdate = dt.date(2025, 5, 18)

print(userdate)
```

```python
diifDate = currentDate - userdate

print(diifDate)
```

# Time

```python
userTime = dt.time(12, 45, 55)

print(userTime)
```

# Date Time

```python
currentDateTime = dt.datetime.now()

print(currentDateTime)

date = currentDateTime.date()

print(date)

time = currentDateTime.time()

print(time)
```

```python
userDateTime = dt.datetime(2025, 5, 18, 12, 50, 23)

print(userDateTime)
```

# Date Time Format

```python
currentDateTime = dt.datetime.now()

print(currentDateTime)

print('---------------------------------------')

format01 = currentDateTime.strftime("%B")

print(format01)

format02 = currentDateTime.strftime("%d / %b / %Y - %B")

print(format02)
```

# Random Module

```python
import random as rn
```

```python
r1 = rn.randint(1, 10)

print(r1)
```

```python
nameList = ['Sara', 'Lara', 'Ram']

r2 = rn.choice(nameList)

print(r2)
```
