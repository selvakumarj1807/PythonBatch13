# Set
-----

1. Set is Collection of Data Type.
2. Declare - { } 
3. It can store multiple values.different data type values in a single variable. Based on non Index Position.
4. Set is Mutable(Changeable)
5. Remove Douplicate Values

```python
userData = {10, 'Sara', 25, 'male', 'Sara', 25}

print(userData, type(userData))
```

# Add Value

```python
userData = {10, 'Sara', 25, 'male', 'Sara', 25}

print(userData)

print('-----------------------------------')

userData.add(9500912258)

print('-----------------------------------')

print(userData)
```

# Update

```python
userData01 = {10, 'Sara', 25, 'male', 'Sara', 25}

userData02 = {9500912258, 'TN'}

print('-----------------------------------')

userData01.update(userData02)

print('-----------------------------------')

print(userData01)
```

# Remove Value USE Remove()

```python
userData = {10, 'Sara', 25, 'male', 'Sara', 25}

print(userData)

print('-----------------------------------')

userData.remove('male')

print('-----------------------------------')

print(userData)

print('-----------------------------------')

userData.remove(9500912258)

print(userData)
```

# Remove Value USE Discard()

```python
userData = {10, 'Sara', 25, 'male', 'Sara', 25}

print(userData)

print('-----------------------------------')

userData.discard('male')

userData.discard(9500912258)

print('-----------------------------------')

print(userData)
```

# Set Operators

```python
v1 = {1, 2, 3, 4, 5, 2, 3, 6}

v2 = {4, 5, 6, 7, 8, 1}
```

```python
# Union

result01 = v1.union(v2)

print(result01)
```

```python
# Intersect

result02 = v1.intersection(v2)

print(result02)
```

```python
# Symmetric Difference

result03 = v1.symmetric_difference(v2)

print(result03)
```
