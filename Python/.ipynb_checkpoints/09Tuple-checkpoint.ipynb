# Tuple
-------

1. Tuple is Collection of Data Type.
2. Declare - ( )
3. It can store multiple values.different data type values in a single variable. Based on Index Position.
4. Tuple is ImMutable(Un Changeable)

```python
userData = (101, 'Sara', 25, 9500912258)

print(userData, type(userData))

print(userData[1])
```

```python
userData[1] = 'Dhara'
```

```python
userData.append(10)
```

# Type Casting Possible (Covert One Data Type to Another Data Type)

Int - Float

Float - Int

Int, Float - String

String - Int, Float (Not Possible)

List - Tuple

Tuple - List

List, Tuple - Set

Set - List, Tuple


# Basic Type Casting

```python
a = 90

b = 95.2

print(a, ' - ', b)

print(float(a), ' - ', int(b))
```
