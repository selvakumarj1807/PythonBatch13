# Python Functions

# 01 No Return Type With Out Arguments

```python
# Create the Function

def msg():
    print('Welcome Python Functions!....')
```

```python
# Call the Function

msg()

print('--------------------------------')

msg()

print('--------------------------------')

msg()
```

```python
# Create the Function

def addTwoNum():
    a = int(input('Enter value 01 : '))
    b = int(input('Enter value 02 : '))
    c = a + b
    
    print(a, '+', b, '=', c)
```

```python
# Call the Function

addTwoNum()

print('--------------------------------')

addTwoNum()
```

# 02 No Return Type With Arguments

```python
# Create the Function

def addTwoNum(a, b):
    c = a + b
    
    print(a, '+', b, '=', c)
```

```python
# Call the Function

addTwoNum(10, 2)

print('--------------------------------')

addTwoNum(20, 4)
```

# 03 Return Type With Out Arguments

```python
# Create the Function

def addTwoNum():
    a = int(input('Enter value 01 : '))
    b = int(input('Enter value 02 : '))
    c = a + b
    
    return c
```

```python
# Call the Function

print('Print Statement : ', addTwoNum())

print('--------------------------------')

result01 = addTwoNum()

print('Call the Function in Variable : ', result01)
```

# 04 Return Type With Arguments

```python
# Create the Function

def addTwoNum(a, b):
    c = a + b
    
    return c
```

```python
# Call the Function

print('Print Statement : ', addTwoNum(12, 2))

print('--------------------------------')

result01 = addTwoNum(22, 4)

print('Call the Function in Variable : ', result01)
```

# Aribitrary Argument Function

```python
# Create the Function

def studentData(*data):
    print(data, type(data))
```

```python
# Call the Function

studentData('Sara', 25, 'Female', 9500912258)
```

# Pass List Values in Argument

```python
# Create the Function

def studentData(data):
    print(data, type(data))
```

```python
# Call the Function

studentData(['Sara', 25, 'Female', 9500912258])
```

# Key Word Argument Function

```python
# Create the Function

def studentData(name, age):
    print('Hi', name, 'your age is', age)
```

```python
# Call the Function

studentData('Sara', 25)

studentData(age = 27, name = 'Lara')
```

# Key Word Arbitrary Argument Function

```python
# Create the Function

def studentData(**data):
    print(data, type(data))
```

```python
# Call the Function

studentData(age = 27, name = 'Lara', mobile = 9500912258)
```

# Default Parametor Argument Function

```python
# Create the Function

def studentData(name, age, state = 'TN'):
    print('Hi', name, 'your age is', age, 'Your state is', state)
```

```python
# Call the Function

studentData('Sara', 25, 'KL')

studentData('Sara', 25)
```

# Lambda Function

```python
addTwoNum = lambda a, b: a + b

print(addTwoNum(10, 2))

print(addTwoNum(20, 2))
```

```python
addTwoNum = lambda a, b: f"Addition of Two Numbers : {a} + {b} = {a + b}"

print(addTwoNum(10, 2))

print(addTwoNum(20, 2))
```
