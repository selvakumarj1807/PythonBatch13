# 1. Arithmatic Operators

```python
a = 7

b = 2

print('Addition (+): ', a + b)

print('Subtraction (-): ', a -b)

print('Multiplication (*): ', a * b)

print('Divsion (/): ', a / b) # It will return Qutiont Value

print('-----------------------------------')

print('Modulas (%): ', a % b) # It will return Remainder Value

print('Floor Division (//): ', a // b)

print('Exponentiation (**): ', a ** b) # 7 ** 2 = 7 * 7

print('Exponentiation (**): ', a ** 3) # 7 ** 3 = 7 * 7 * 7
```

# 3. Comparision or RelationOperators

```python
a = 10

b = 2

print(a == b)

print(a != b)

print(a <= b) # 10 <= 2

print(a >= b) # 10 >= 2
```

# 4. Logical Operators

```python
userName = input('Enter your User Name')

password = int(input('Enter your Password'))

print('--------------------------------------------')

print(userName == 'Sara')

print(password == 123)

print('--------------------------------------------')

print((userName == 'Sara') and (password == 123))
```

```python
userName = input('Enter your User Name')

password = int(input('Enter your Password'))

print('--------------------------------------------')

print(userName == 'Sara')

print(password == 123)

print('--------------------------------------------')

print((userName == 'Sara') or (password == 123))
```

# 5. Membership Operators

```python
userData = [10, 'Sara', 25, True]

print(userData)

print('------------------------------------------')

print('Sara' in userData)

print('male' in userData)

print('male' not in userData)
```

# 6. Identity Operators

```python
a = 10

b = 10

print(id(a))

print(id(b))

print(a is b)

print(a is not b)
```

# To Find the Area of the Circle (A=πr^2)

```python

pi = 3.14

r = int(input('Enter your Radius Value : '))

A = pi * r * r

print('Area of the Circle : ', A)
```

# 7. Bitwise Operators

```python
'''
a = 10 = 1010

b = 2  = 0010 # 0 = False, 1 = True - &
-------------
       = 0010
'''

a = 10

b = 2

print('Bitwise and Operators (&): ', a & b)
```

```python
'''
a = 10 = 1010

b = 2  = 0010 # 0 = False, 1 = True - &
-------------
       = 1010
'''

a = 10

b = 2

print('Bitwise or Operators (|): ', a | b)
```

```python
'''
a = 10

~a = -(a + 1)
   = -(10 + 1)
   = -(11)
'''

a = 10

print('Bitwise Not Operators :', ~a)
```

```python
'''
a = 10

a << 2 = 10 << 2
       = 10 * (2 * 2) = 10 * (4)

a << 3 = 10 << 3
       = 10 * (2 * 2 * 2) = 10 * (8)
'''

a = 10

print('Bitwise Left Shift Operators : ', a << 2)

print('Bitwise Left Shift Operators : ', a << 3)
```

```python
'''
a = 10

a >> 2 = 10 >> 2
       = 10 / (2 * 2) = 10 / (4)

a >> 3 = 10 >> 3
       = 10 / (2 * 2 * 2) = 10 / (8)
'''

a = 10

print('Bitwise Right Shift Operators : ', a >> 2)

print('Bitwise Right Shift Operators : ', a >> 3)
```
