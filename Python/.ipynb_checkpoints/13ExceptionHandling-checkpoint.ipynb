# Exception Handling

# try & except

```python
try:
    a = 10

    b = 0
    
    c = a / b
    
    print(c)

except:
    print('Divisor cannot be zero')
```

# Exception Variable

```python
try:
    a = 10

    b = 0
    
    c = a / b
    
    print(c)

except Exception as e:
    print(e)
    print('Divisor cannot be zero')
```

# try & except & else

```python
try:
    a = 10

    b = 2
    
    c = a / b

except Exception as e:
    print(e)

else:
    print(c)
```

# try & except & else & finally

```python
try:
    a = 10

    b = 0
    
    c = a / b

except Exception as e:
    print(e)

else:
    print(c)

finally:
    print('Program Run Successfully!...')
```
