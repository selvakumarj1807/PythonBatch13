# Dictionary
------------

1. Dictionary is Collection of Data Type.
2. Declare - {key: value} 
3. It can store multiple values.different data type values in a single variable. Based on non Index Position.
4. Dictionary is Mutable(Changeable)
5. Remove Douplicate Values

```python
userData = {'id': 10, 'name': 'Sara', 'age': 25}

print(userData, type(userData))
```

```python
print(userData.keys())

print(userData.values())
```

# Get Value use key name

```python
print(userData['name'])
```

# Add Values

```python
userData = {'id': 10, 'name': 'Sara', 'age': 25}

print(userData)

print('-------------------------------------------')

userData.update({'gender': 'male', 'mobile': 9500912258})

print('-------------------------------------------')

print(userData)
```

# Change Value Use Key Name

```python
userData = {'id': 10, 'name': 'Sara', 'age': 25}

print(userData)

print('-------------------------------------------')

userData['name'] = 'Dahara'

print('-------------------------------------------')

print(userData)
```

# Remove Values Use Key name

```python
userData = {'id': 10, 'name': 'Sara', 'age': 25}

print(userData)

print('-------------------------------------------')

userData.pop('age')

print('-------------------------------------------')

print(userData)
```

# Use for Loop 

```python
userData = {'id': 10, 'name': 'Sara', 'age': 25}

print(userData)

print('-------------------------------------------')

for i in userData:
    print(i, ' - ', userData[i])
```

# Dictionary of Dictionary Values

```python
users = {
    'user01': {'id': 10, 'name': 'Sara', 'age': 25},
    'user02': {'id': 20, 'name': 'Lara', 'age': 22},
    'user03': {'id': 30, 'name': 'Dara', 'age': 26}
}

print(users)
```

```python
print(users['user01']['id'])

print(users['user01']['name'])

print(users['user01']['age'])

print('----------------------------------------------')

print(users['user02']['id'])

print(users['user02']['name'])

print(users['user02']['age'])
```

```python
for i in users:
    print(i, ':', users[i]['id'])

    print(i, ':', users[i]['name'])

    print(i, ':', users[i]['age'])

    print('----------------------------------------')
```
