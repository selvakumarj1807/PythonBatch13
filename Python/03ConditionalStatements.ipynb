# If

```python
age = int(input('Enter your Age Value : '))

if age >= 18:
    print(age, ' Is 18 +. So you are eligible for vote')
```

# If & Else

```python

age = input('Enter your Age Value : ')

print(type(age))

if int(age) >= 18:
    print(age, ' Is 18 +. So you are eligible for vote')

else:
    print(age, ' Is 18 -. So you are not eligible for vote')
```

# If Elif Else

```python
m1 = int(input('Enter mark 01 : '))
m2 = int(input('Enter mark 02 : '))
m3 = int(input('Enter mark 03 : '))

total = m1 + m2 + m3

avg = total // 3

if avg >= 90 and avg <= 100: 
    print(avg, 'A Grade')
    
elif avg >= 80 and avg <= 90:
    print(avg, 'B Grade')

elif avg >= 70 and avg <= 80:
    print(avg, 'B Grade')

elif avg >= 60 and avg <= 70:
    print(avg, 'B Grade')

else:
    print(avg, 'No Grade')
```

# Nested If
