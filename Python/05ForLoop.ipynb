# While Loop

```python
i = 0

while i <= 5:
    print(i)

    i = i + 1
```

# For Loop

Default Value Initial is 0

Default Value Increment is 1

```python
for i in range(5):
    print(i)
```

# Odd Number

```python
for i in range(1, 10, 2):
    print(i)
```

# Even Number

```python
for i in range(0, 10, 2):
    print(i)
```

# Nested Loop

```python
for i in range(1, 6):
    for j in range(i):
        print(i, end=' ')

    print('')
```

```python
for i in range(1, 6):
    for j in range(1, i+1):
        print(j, end=' ')

    print('')
```

```python
for i in range(1, 6):
    for j in range(i):
        print('#', end=' ')

    print('')
```
