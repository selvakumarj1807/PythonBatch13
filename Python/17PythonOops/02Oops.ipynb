# Create Function Inside the Class

```python
class Student:
    name = 'Dharani'
    age = 27

    def details():
        print(Student.name, 'your age is', Student.age)
```

# Call the Function From the Class

```python
# Dot Notation

Student.details()

# get Attribute

getattr(Student,'details')()
```

# Create Function Inside the Class

```python
class Student:
    name = 'Student Name'
    age = 'Student Age'

    def details(self):
        print(self.name, 'your age is', self.age)
```

```python
# Create the Object

studentOject01 = Student()
```

# Call the Function From the Object

```python
studentOject01.name = 'Sam'

studentOject01.age = 27

studentOject01.details()
```

```python
# Create the Object

studentOject02 = Student()

studentOject02.name = 'Ram'

studentOject02.age = 25

studentOject02.details()
```

# Init Method

print('Constructor method for Value Initial!...')

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def details(self):
        print(self.name, 'your age is', self.age)
```

```python
# Create the Object

userObject01 = User('Dharani', 27)

userObject01.details()
```

```python
# Create the Object

userObject02 = User('Priya', 22)

userObject02.details()
```
