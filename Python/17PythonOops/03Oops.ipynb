# Inheritance

# 01 Single Inheritance

```python
class Father:
    def ownBike(self):
        print('Father have Own Bike')

class Son(Father):
    def pen(self):
        print('Son Have One Pen')

object01 = Son()
```

```python
object01.pen()

object01.ownBike()
```

# 02 Multiple Inheritance

```python
class Father:
    def ownBike(self):
        print('Father have Own Bike')

    def playing(self):
        print('Father have Playing Skill')

class Mother:
    def cooking(self):
        print('Mother have Cooking Skill')

    def playing(self):
        print('Mother have Playing Skill')

class Son(Mother, Father):
    def pen(self):
        print('Son Have One Pen')

object01 = Son()
```

```python
object01.pen()

object01.ownBike()

object01.cooking()

object01.playing()
```

# 03 Multi Level Inheritance

```python
class Father:
    def ownBike(self):
        print('Father have Own Bike')

    def playing(self):
        print('Father have Playing Skill')

class Mother(Father):
    def cooking(self):
        print('Mother have Cooking Skill')

    def playing(self):
        print('Mother have Playing Skill')

class Son(Mother):
    def pen(self):
        print('Son Have One Pen')

object01 = Son()
```

```python
object01.pen()

object01.ownBike()

object01.cooking()

object01.playing()
```
