```python
# Create the Class

class Library:
    def __init__(self, books):
        self.books = books

    def displayBooks(self):
        for i in self.books:
            print(i, end=', ')

    def borrowBook(self, book):
        if book in self.books:
            print('Get Your', book ,'Book Now!...')
            self.books.remove(book)

        else:
            print(book, 'Book Not in your Library!...')

    def reciveBook(self, book):
        self.books.append(book)
        print(book, ' - Book Recive Library Successfully!...')
        
```

```python
# print('Library Management System')

books = ['C', 'C++', 'Java']

library01 = Library(books)

msg = '''
1. Display Books
2. Borrow Book
3. Recive Book
'''

while True:
    print(msg)

    choice = input('Enter Your Choice (1/2/3) :')
    
    if choice == '1':
        print('Our Library Books!...')

        library01.displayBooks()
        print('')

    elif choice == '2':
        book = input('Enter your Borrow Book Name : ')

        library01.borrowBook(book)
        print('')

    elif choice == '3':
        book = input('Enter your Recive Book Name : ')
        
        library01.reciveBook(book)
        print('')

        
    else:
        print('Enter a Valid Choice!...')
        break
```

```python
# print('Library Management System')

books = ['Tamil', 'English', 'Maths', 'Sci']

library02 = Library(books)

msg = '''
1. Display Books
2. Borrow Book
3. Recive Book
'''

while True:
    print(msg)

    choice = input('Enter Your Choice (1/2/3) :')
    
    if choice == '1':
        print('Our Library Books!...')

        library02.displayBooks()
        print('')

    elif choice == '2':
        book = input('Enter your Borrow Book Name : ')

        library02.borrowBook(book)
        print('')

    elif choice == '3':
        book = input('Enter your Recive Book Name : ')
        
        library02.reciveBook(book)
        print('')

        
    else:
        print('Enter a Valid Choice!...')
        break
```
