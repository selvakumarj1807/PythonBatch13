# Create the Class

```python
class Student:
    name = 'Ram'
    age = 25
```

# Call the Class Variable

```python
# Use Dot Notation

print(Student.name)

# Use Get Attribute

print('Age : ', getattr(Student, 'age'))
```

# Create the Class

```python
class User:
    pass
```

# Create the Class Variable. After Class Creatioon

```python
# Use Dot Notaion

User.name = 'Sam'

print(User.name)

# Use Set Attribute

setattr(User, 'age', 30)

print('Age :', User.age)
```

# Create the Class

```python
class Student:
    name = 'Ram'
    age = 25
```

# Delete the Class Variable. After Class Creation

```python
# Use Dot Notation

print(Student.name)

del(Student.name)

print(Student.name)
```

```python
# User DelAttribute

print(Student.age)

delattr(Student, 'age')

print(Student.age)
```
