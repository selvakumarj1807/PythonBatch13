# List
------

1. List is Collection of Data Type.
2. Declare - [ ]
3. It can store multiple values.different data type values in a single variable. Based on Index Position.
4. List is Mutable(Changeable)

```python
userData = [101, 'Sara', 25, 'male', True]

print(userData, type(userData))
```

```python
userData = [101, 'Sara', 25, 'male', True]

print(userData)

print('-----------------------------------------')

userName = userData[1]

print(userName, type(userName))

print('-----------------------------------------')

print(userData[1:4])
```

# Conditional Statements

```python
# IF

userData = [101, 'Sara', 25, 'male', True]

print(userData)

print('-----------------------------------------')

if 'Sara' in userData:
    print('Value is There')
else:
    print('Value Not in Your Variable')
```

```python
# For Loop

userData = [101, 'Sara', 25, 'male', True]

print(userData)

print('-----------------------------------------')

for i in userData:
    print(i)
```

# List of List Values

```python
userData = [101, 'Sara', 25, [95, 90, 95]]

print(userData)

print('-----------------------------------------')

userName = userData[1]

print(userName, type(userName))

print('-----------------------------------------')

studentMarks = userData[3]

print(studentMarks, type(studentMarks))

print('-----------------------------------------')

print(studentMarks[1])

print(userData[3][1])
```

# Append

```python
userData = [101, 'Sara', 25]

print(userData)

print('-----------------------------------------')

userData.append("St Joseph's College")

print('-----------------------------------------')

print(userData)
```

# Change Existing value

```python
userData = [101, 'Sara', 25]

print(userData)

print('-----------------------------------------')

userData[1] = 'Dharani'

print('-----------------------------------------')

print(userData)
```

# Extend

```python
userData01 = [101, 'Sara', 25]

print(userData01)

userData02 = ['male', True]

print(userData02)

print('-----------------------------------------')

userData01.extend(userData02)

print('-----------------------------------------')

print(userData01)
```

# Delete Value Based on Value

```python
userData = [101, 'Sara', 25]

print(userData)

print('-----------------------------------------')

userData.remove(25)

print('-----------------------------------------')

print(userData)
```

# Delete Value based on Index Position

```python
userData = [101, 'Sara', 25]

print(userData)

print('-----------------------------------------')

userData.pop(2)

print('-----------------------------------------')

print(userData)
```

# Insert

```python
userData = [101, 'Sara', 25]

print(userData)

print('-----------------------------------------')

userData.insert(2, 'male')

print('-----------------------------------------')

print(userData)
```

# Sorting

```python
ageValues = [18, 20, 17, 28, 25, 22, 21, 30, 27, 20]

print(ageValues)

print('-----------------------------------------')

ageValues.sort()

print('-----------------------------------------')

print(ageValues)

print('-----------------------------------------')

ageValues.sort(reverse=True)

print('-----------------------------------------')

print(ageValues)
```
