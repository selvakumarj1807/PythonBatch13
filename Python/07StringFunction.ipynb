# Strip, lStrip, rStrip

```python
# Strip

msg = '##95#00##'

print(msg)

print('----------------------------')

print(msg.strip('#'))

print('----------------------------')

print(msg.lstrip('#'))

print('----------------------------')

print(msg.rstrip('#'))

print('----------------------------')

msg = '  Hi  '

print(msg, len(msg))

print(msg.strip(), len(msg.strip()))
```

# Split Lines

```python
para = """Yes, Luxembourg is significantly richer than India, especially when measured by GDP per capita,
which reflects individual prosperity, with Luxembourg consistently ranking as one of the world's wealthiest nations per person,
while India, despite its large overall economy, has a much lower per capita income due to its massive population. 
Luxembourg's high GDP per capita (often exceeding $150,000) reflects high living standards,
while India's lower figure (around $2,400-$3,000) shows a significant gap in average wealth, 
even though India has a top 5 total GDP globally. 
"""

print(para)

print('-------------------------------------------------------------------')

v2 = para.splitlines()

print(v2)
```
